```python
import omnigibson as og

og.launch()

from omnigibson.utils.asset_utils import get_all_object_category_models_with_abilities
```

```python
from bddl.knowledge_base import *
```

```python
tro_synsets = {
    s for s in Synset.all_objects() if len(s.tasks) > 0 and s.state != STATE_SUBSTANCE
}
```

```python
from bddl.object_taxonomy import ObjectTaxonomy

OT = ObjectTaxonomy()
```

```python
no_good_obj = {
    s
    for s in tro_synsets
    if all(
        len(
            get_all_object_category_models_with_abilities(
                c.name, OT.get_abilities(c.synset.name)
            )
        )
        == 0
        for c in {obj.category for obj in s.matching_objects}
    )
}
sorted(no_good_obj)
```
