```python
import glob, pathlib, json

files = [
    pathlib.Path(x)
    for x in glob.glob(
        "/scr/BEHAVIOR-1K/asset_pipeline/cad/scenes/*/artifacts/object_list.json"
    )
]
data = {p.parts[-3]: json.loads(p.read_text()) for p in files}
```

```python
in_portals = [
    (k + "-in", v["incoming_portal"]) for k, v in data.items() if v["incoming_portal"]
]
```

```python
out_portals = [
    (k + f"-out-{sc}-{i}", op)
    for k, v in data.items()
    for i, (sc, op) in enumerate(v["outgoing_portals"].items())
]
```

```python
portals = in_portals + out_portals
```

```python
portals
```

```python
portal_rots = {n: p[1] for n, p in portals}
```

```python
portal_rots
```

```python
from scipy.spatial.transform import Rotation as R
import numpy as np

rots = {k: R.from_quat(v) for k, v in portal_rots.items()}
bad_rots = {
    k: r.as_rotvec()
    for k, r in rots.items()
    if r.magnitude() > 1e-3
    and not np.allclose(np.abs(r.as_rotvec() / r.magnitude()), [0, 0, 1], atol=1e-3)
}
```
