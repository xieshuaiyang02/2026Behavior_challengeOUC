```python
import json, pathlib
```

```python
by_target = {}
for p in pathlib.Path(r"D:\BEHAVIOR-1K\asset_pipeline").glob(
    "cad/*/house_*/artifacts/check_collisions.json"
):
    with p.open() as f:
        target = "/".join(p.parts[-4:-2])
        by_target[target] = json.load(f)
```

```python
from bddl.knowledge_base import *
```

```python
for (from_cat, from_mdl, from_inst), (to_cat, to_mdl, to_inst), dist in by_target[
    "scenes/house_double_floor_lower"
]["collisions"]:
    # Is either model a door? Skip if so
    # if Object.get(from_mdl).category.name == "door" or Object.get(to_mdl).category.name == "door":
    #     continue
    print(
        f"Collision between {from_cat}-{from_mdl}-{from_inst} and {to_cat}-{to_mdl}-{to_inst}: {dist:.2f}mm"
    )
```

```python
# Find out the errors
for target, data in by_target.items():
    err = data["error"]
    if not err:
        continue
    print(target, err)
```

```python
# Check the number of collisions per file and total
collisions = {t: len(data["collisions"]) for t, data in by_target.items()}
print(*sorted(collisions.items(), key=lambda x: x[1], reverse=True), sep="\n")
print("Total", sum(collisions.values()))
```
