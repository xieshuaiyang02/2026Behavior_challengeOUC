```python
import xml.etree.ElementTree as ET
import numpy as np
from scipy.spatial.transform import Rotation as R

FILE_PATH = r"C:\Users\cgokmen\Downloads\scenes\scenes\Beechwood_0_garden\urdf\Beechwood_0_garden_best.urdf"

# Get the joint axes in the object
with open(FILE_PATH) as f:
    tree = ET.parse(f)
joints = list(tree.findall(".//joint"))

states = {}
for joint in joints:
    name = joint.attrib["name"].replace("j_", "").replace("-", "_")
    position = np.array(
        [float(x) for x in joint.find("origin").attrib["xyz"].split(" ")]
    )
    orientation = np.array(
        [float(x) for x in joint.find("origin").attrib["rpy"].split(" ")]
    )
    states[name] = (position, R.from_euler("xyz", orientation).as_quat())
```
