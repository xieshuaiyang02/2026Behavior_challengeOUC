```python
import pathlib, shutil
from tqdm.notebook import tqdm
```

```python
images = list(
    pathlib.Path(
        "/scr/BEHAVIOR-1K/asset_pipeline/artifacts/pipeline/object_images/"
    ).glob("*.webp")
)
targets = [
    pathlib.Path(
        "/scr/BEHAVIOR-1K/asset_pipeline/artifacts/pipeline/object_images/renamed"
    )
    / pathlib.Path(p.stem.split("-")[-1]).with_suffix(".webp")
    for p in images
]
```

```python
for i, t in tqdm(zip(images, targets)):
    shutil.copy2(i, t)
```
