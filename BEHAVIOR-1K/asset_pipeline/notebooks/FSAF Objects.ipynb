```python
from bddl.knowledge_base import *
```

```python
objs = [
    obj
    for obj in Object.all_objects()
    if any(ml.name == "togglebutton" for ml in obj.meta_links)
]
```

```python
interesting_synset_names = [
    "sink.n.01",
    "stove.n.01",
    "burner.n.02",
    "microwave.n.02",
    "oven.n.01",
]  # ["appliance.n.02", "fixture.n.01"]
interesting_synsets = [Synset.get(name) for name in interesting_synset_names]
assert all(interesting_synsets)
interesting_synset_objects = [
    obj
    for obj in objs
    if any(
        obj.category.synset == interesting_synset
        or interesting_synset in obj.category.synset.ancestors
        for interesting_synset in interesting_synsets
    )
]
interesting_synset_objects
```

```python
import json

interesting_objects = [
    (obj.category.name, obj.name) for obj in interesting_synset_objects
]
with open("fsaf_objects.json", "w") as f:
    json.dump(sorted(interesting_objects), f, indent=2)
```
