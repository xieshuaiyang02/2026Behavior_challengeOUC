```python
import glob, os
```

```python
bddls = {}
for fn in glob.glob(r"/scr/bddl/bddl/activity_definitions/*/*.bddl"):
    bn = os.path.basename(os.path.dirname(fn))
    with open(fn, "r") as f:
        lines = f.read()
        bddls[bn] = lines
```

```python
len(bddls)
```

```python
# Contains one of the problematic states?
problematic_states = {"attached", "assembled"}
problematic_bddls = {
    k: v for k, v in bddls.items() if any(x in v for x in problematic_states)
}
```

```python
len(problematic_bddls)
```

```python
problematic_bddls_details = []
for name, bddl in problematic_bddls.items():
    problematic_lines = "\n".join(
        [l for l in bddl.split("\n") if any(x in l for x in problematic_states)]
    )
    problematic_bddls_details.append(
        {
            "name": name,
            "all_lines": bddl,
            "attach_lines": problematic_lines,
            "todo": "",
            "type": "",
            "notes": "",
        }
    )
```

```python
import pandas as pd

df = pd.DataFrame.from_records(problematic_bddls_details)
print(df)
```
